```python
from sklearn.datasets import load_iris
import seaborn as sns
from sklearn.preprocessing import StandardScaler
```

```python
iris = load_iris()

X = iris.data
y= iris.target
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(X)
```

```python
sns.scatterplot(x=x_scaled[:,0],y=x_scaled[:,2])
```

```python
# dbscan
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(eps=0.5,min_samples=5)
labels = dbscan.fit_predict(x_scaled)
```

```python
sns.scatterplot(x=x_scaled[:,0],y=x_scaled[:,2],c=labels)
```

# Non-linear data

```python
from sklearn.datasets import make_moons

X,y = make_moons(
    n_samples = 300,
    noise = 0.05,
    random_state = 42
)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
sns.scatterplot(x=X_scaled[:,0],y=X_scaled[:,1])
```

# k means

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=2,
                random_state=42)

labels = kmeans.fit_predict(X_scaled)
```

```python
sns.scatterplot(x=X_scaled[:,0],y=X_scaled[:,1],c=labels)
```

```python
# applying dbscan on moons dataset

dbscan = DBSCAN(eps=0.2,min_samples=5)
labels = dbscan.fit_predict(X_scaled)
```

```python
sns.scatterplot(x=X_scaled[:,0],y=X_scaled[:,1],hue=labels)
```
