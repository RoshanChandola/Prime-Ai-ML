# Anomaly Detection using DBSCAN and other Algo

```python
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons
```

```python
X,y = make_moons(
    n_samples = 500,
    noise = 0.1,
    random_state =42
)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# visulaize data
sns.scatterplot(x = X_scaled[:,0], y = X_scaled[:,1])
```

```python
# DBSCAN model
# here the most crucial part is to choose min_samples and eps. If you choose a very small value for eps, then the model will consider most of the points as outliers. If you choose a very large value for eps, then the model will consider most of the points as inliers. So, it is very important to choose the right value for eps and min_samples. so we can use the elbow method to find the optimal value for eps. The elbow method is a graphical method that helps to find the optimal value for eps. The idea is to plot the number of clusters formed by the model against the value of eps. The point where the number of clusters starts to decrease rapidly is called the elbow point. The value of eps at this point is considered as the optimal value for eps.

#lets choose the value using the elbow method. We can use the following code to find the optimal value for eps.

dbscan = DBSCAN(
    min_samples=4,
    eps=0.2
    )

labels = dbscan.fit_predict(X_scaled)

sns.scatterplot(x = X_scaled[:,0], y = X_scaled[:,1], hue=labels, palette='Set1')
```
